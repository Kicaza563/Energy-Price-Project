# Section 6: Outliers

The final measurement of GDP Deflator at 2022 are listed as 1. This is possibly a measurement error, as a 0.08 jump in one year is too much and is likely just caused by rounding.

```python
price_cols = df.columns.drop(['Year', 'Sector', 'GDP Deflator'])
rows = []
for col in price_cols :
    # Strip out any string characters like '$' and ',' then convert to numeric
    cleaned_series = df[col].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False)
    numeric_series = pd.to_numeric(cleaned_series, errors='coerce')
    s = numeric_series.dropna()

    if len(s) == 0:
        continue

    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    is_outlier = (numeric_series < q1 - 1.5 * iqr) | (numeric_series > q3 + 1.5 * iqr) # the usual 1.5 x IQR rule
    for i in df.index[is_outlier] :
      rows.append({'Feature': col, 'Year': df.loc[i, 'Year'], 'Sector': df.loc[i, 'Sector'], 'Value': df.loc[i, col]})

outliers = pd.DataFrame(rows).sort_values(['Feature', 'Year'])
print(outliers.groupby('Feature').size())
```
